```python
from sklearn.svm import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
```

```python
df=datasets.load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
x=df.drop(columns="target")
y=df["target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
scaler=StandardScaler()
y_train_scaled=scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled=scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
y_train_scaled
```

```python
model=SVR()
model.fit(x_train,y_train_scaled)
```

```python
y_pred=model.predict(x_test)
y_train_pred=model.predict(x_train)
```

```python
print("r2_score:",r2_score(y_train_scaled,y_train_pred))
print("r2_score:",r2_score(y_test_scaled,y_pred))
```

```python
model=SVR(kernel="sigmoid")
model.fit(x_train,y_train_scaled)
y_pred=model.predict(x_test)
print("r2_score:",r2_score(y_test_scaled,y_pred))
```

```python
from sklearn.model_selection import GridSearchCV
param_grid={
    "C":[1,2,10,20,50],
    "kernel":["rbf","linear"],
    "epsilon":[0.01,0.1,0.2,0.5,1.0]
}

 
```

```python
svr=SVR()
grid_search= GridSearchCV(svr,param_grid,scoring="r2",cv=5)
```

```python
grid_search.fit(x_train,y_train_scaled)
```

```python
print("best_param-",grid_search.best_params_)
```

```python
best_model=SVR(kernel='linear',C=10,epsilon=0.1)
best_model.fit(x_train,y_train_scaled)
y_pred=best_model.predict(x_test)
y_train_pred=best_model.predict(x_train)
print("r2_score:",r2_score(y_train_scaled,y_train_pred))
print("r2_score:",r2_score(y_test_scaled,y_pred))
```
